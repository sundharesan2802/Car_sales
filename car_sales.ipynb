```python
#10 Month by Month Car sales trend

sales["Date"] = pd.to_datetime(sales["Date"])

monthly_sales = (
    sales.groupby(sales["Date"].dt.to_period("M"))
    .size()
    .reset_index(name="Sales Count")
)

monthly_sales["Date"] = monthly_sales["Date"].astype(str)

monthly_sales
```

```python
# Highest Total sales Revenue by companies

company_revenue = (
    sales.groupby("Company")["Price ($)"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)




# Top 10 best car selling models

top_models = sales["Model"].value_counts().head(10)

8



# Top 5 companies total sales percentage

company_sales = sales["Company"].value_counts()

top5_sales = company_sales.head(5).sum()
total_sales = company_sales.sum()

top5_percentage = (top5_sales / total_sales) * 100

top5_percentage

sales_share = [
    top5_sales,
    total_sales - top5_sales
]



# Top 10 companies unique model prices

top10_companies = sales["Company"].value_counts().head(10).index

unique_models = (
    sales[sales["Company"].isin(top10_companies)]
    .groupby(["Company", "Model"])["Price ($)"]
    .mean()
    .reset_index()
)

```

```python
# DashBoard


fig, axes = plt.subplots(2, 2, figsize=(18, 12))


# Chart 1
sns.barplot(
    x=company_revenue.values,
    y=company_revenue.index,
    hue=company_revenue.index,
    palette="viridis",
    legend=False,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Top 10 Companies by Total Sales Revenue", fontweight="bold")
axes[0, 0].set_xlabel("Total Revenue ($)")
axes[0, 0].set_ylabel("Company")
axes[0, 0].ticklabel_format(style="plain", axis="x")

for container in axes[0, 0].containers:
    axes[0, 0].bar_label(
        container,
        fmt="$%.0f",
        label_type="center"
    )


# Chart 2
sns.barplot(
    x=top_models.values,
    y=top_models.index,
    hue=top_models.index,
    palette="coolwarm",
    legend=False,
    ax=axes[0, 1]
)

axes[0, 1].set_title("Top 10 Best-Selling Car Models", fontweight="bold")
axes[0, 1].set_xlabel("Sales Count")
axes[0, 1].set_ylabel("Model")

for container in axes[0, 1].containers:
    axes[0, 1].bar_label(
        container,
        fmt="%d",
        label_type="center"
    )


# Chart 3
axes[1, 0].pie(
    sales_share,
    labels=["Top 5 Companies", "Other Companies"],
    autopct="%1.1f%%",
    startangle=90,
    colors=["skyblue", "lightcoral"],
    explode=[0.05, 0]
)

axes[1, 0].set_title(
    "Top 5 Companies Share of Total Sales",
    fontweight="bold"
)


# Chart 4
sns.stripplot(
    data=unique_models,
    x="Company",
    y="Price ($)",
    hue="Company",
    palette="Set2",
    legend=False,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Unique Model Prices of Top 10 Companies",
    fontweight="bold"
)

axes[1, 1].set_xlabel("Company")
axes[1, 1].set_ylabel("Average Price ($)")
axes[1, 1].tick_params(axis="x", rotation=45)


plt.tight_layout()
plt.show()
```

```python

#  Total sales revenue by dealer region
region_revenue = (
    sales.groupby("Dealer_Region")["Price ($)"]
    .sum()
    .sort_values(ascending=False)
)


#  Most sold body style in each dealer region
body_region = (
    sales.groupby(["Dealer_Region", "Body Style"])
    .size()
    .reset_index(name="Sales Count")
)

top_body_region = (
    body_region.loc[
        body_region.groupby("Dealer_Region")["Sales Count"].idxmax()
    ]
)


#  Percentage distribution of car sales across dealer regions
region_sales = sales["Dealer_Region"].value_counts()

region_percentage = (
    region_sales / region_sales.sum()
) * 100


#  Top 10 dealers by number of cars sold
top_dealers = sales["Dealer_Name"].value_counts().head(10)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(19, 13))


# ==================== CHART 1 ====================
# Dealer Region total Revenue

sns.barplot(
    x=region_revenue.values,
    y=region_revenue.index,
    hue=region_revenue.index,
    palette="Set3",
    legend=False,
    ax=axes[0, 0]
)

axes[0, 0].set_title(
    "Total Sales Revenue by Dealer Region",
    fontweight="bold",
    fontsize=14
)
axes[0, 0].set_xlabel("Total Revenue ($)")
axes[0, 0].set_ylabel("Dealer Region")
axes[0, 0].ticklabel_format(style="plain", axis="x")

for p in axes[0, 0].patches:
    if p.get_width() > 0:
        axes[0, 0].text(
            p.get_width() / 2,
            p.get_y() + p.get_height() / 2,
            f"${p.get_width():,.0f}",
            ha="center",
            va="center",
            fontsize=9,
            fontweight="bold"
        )


# ==================== CHART 2 ====================
# Most Sold Body Style in Each Region

sns.barplot(
    data=top_body_region,
    x="Dealer_Region",
    y="Sales Count",
    hue="Body Style",
    palette="pastel",
    ax=axes[0, 1]
)

axes[0, 1].set_title(
    "Most Sold Body Style by Dealer Region",
    fontweight="bold",
    fontsize=14
)
axes[0, 1].set_xlabel("Dealer Region")
axes[0, 1].set_ylabel("Sales Count")

for container in axes[0, 1].containers:
    axes[0, 1].bar_label(
        container,
        fmt="%.0f",
        padding=3,
        fontsize=9
    )

axes[0, 1].legend(
    title="Body Style",
    fontsize=8
)


# ==================== CHART 3 ====================
# Regional Sales Percentage Distribution

axes[1, 0].pie(
    region_percentage.values,
    labels=region_percentage.index,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.80,
    explode=[0.03] * len(region_percentage),
    wedgeprops={"width": 0.45, "edgecolor": "white"}
)

axes[1, 0].set_title(
    "Percentage Distribution of Car Sales by Region",
    fontweight="bold",
    fontsize=14
)


# ==================== CHART 4 ====================
# Top 10 Dealers by Cars sold

sns.barplot(
    x=top_dealers.values,
    y=top_dealers.index,
    hue=top_dealers.index,
    palette="Set3",
    legend=False,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Top 10 Dealers by Cars Sold",
    fontweight="bold",
    fontsize=14
)
axes[1, 1].set_xlabel("Cars Sold")
axes[1, 1].set_ylabel("Dealer", labelpad=20)

for p in axes[1, 1].patches:
    if p.get_width() > 0:
        axes[1, 1].text(
            p.get_width() / 2,
            p.get_y() + p.get_height() / 2,
            f"{int(p.get_width())}",
            ha="center",
            va="center",
            fontsize=9,
            fontweight="bold"
        )


plt.tight_layout()
plt.show()
```

```python
#

#  Body style preference by Male and Female
gender_body = (
    sales.groupby(["Gender", "Body Style"])
    .size()
    .reset_index(name="Sales Count")
)


# Most commonly sold engine types
engine_sales = sales["Engine"].value_counts().reset_index()
engine_sales.columns = ["Engine", "Sales Count"]


# Car price Distribution by different body style
body_style_price = (
    sales.groupby("Body Style")["Price ($)"]
    .agg(["min", "max", "mean"])
    .reset_index()
)

body_style_price


#  Car price variation between Auto and Manual
transmission_price = (
    sales.groupby("Transmission")["Price ($)"]
    .mean()
    .reset_index(name="Average Price")
)


#  Quarterly revenue from 2022 to 2023
sales["Date"] = pd.to_datetime(sales["Date"])

quarterly_revenue = (
    sales[sales["Date"].dt.year.isin([2022, 2023])]
    .groupby([
        sales["Date"].dt.year.rename("Year"),
        sales["Date"].dt.quarter.rename("Quarter")
    ])["Price ($)"]
    .sum()
    .reset_index(name="Revenue")
)

quarterly_revenue["Quarter"] = "Q" + quarterly_revenue["Quarter"].astype(str)


#  Monthly car sales
monthly_sales = (
    sales.groupby(sales["Date"].dt.month)["Car_id"]
    .count()
    .reset_index(name="Sales Count")
)

monthly_sales["Month"] = pd.to_datetime(
    monthly_sales["Date"],
    format="%m"
).dt.month_name()

monthly_sales = monthly_sales.sort_values("Date")
```

```python
fig, axes = plt.subplots(3, 2, figsize=(20, 20))


# ==================== CHART 1 ====================
# Male vs Female Body Style Preference

sns.barplot(
    data=gender_body,
    x="Body Style",
    y="Sales Count",
    hue="Gender",
    palette="Set2",
    ax=axes[0, 0]
)

axes[0, 0].set_title(
    "Body Style Preference by Gender",
    fontweight="bold"
)
axes[0, 0].set_xlabel("Body Style")
axes[0, 0].set_ylabel("Sales Count")

for container in axes[0, 0].containers:
    axes[0, 0].bar_label(container, fmt="%.0f", padding=2)


# ==================== CHART 2 ====================
# Engine Type Sales

sns.barplot(
    data=engine_sales,
    x="Engine",
    y="Sales Count",
    hue="Engine",
    palette="pastel",
    legend=False,
    ax=axes[0, 1]
)

axes[0, 1].set_title(
    "Most Commonly Sold Engine Types",
    fontweight="bold"
)
axes[0, 1].set_xlabel("Engine Type")
axes[0, 1].set_ylabel("Sales Count")

for container in axes[0, 1].containers:
    axes[0, 1].bar_label(container, fmt="%.0f", padding=3)


# ==================== CHART 3 ====================
# Price Distribution by Body Style

sns.boxplot(
    data=sales,
    x="Body Style",
    y="Price ($)",
    hue="Body Style",
    palette="Set3",
    legend=False,
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    "Car Price Distribution by Body Style",
    fontweight="bold"
)
axes[1, 0].set_xlabel("Body Style")
axes[1, 0].set_ylabel("Price ($)")
axes[1, 0].ticklabel_format(style="plain", axis="y")


# ==================== CHART 4 ====================
# Average Price - Automatic vs Manual

sns.barplot(
    data=transmission_price,
    x="Transmission",
    y="Average Price",
    hue="Transmission",
    palette="Paired",
    legend=False,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Average Price by Transmission",
    fontweight="bold"
)
axes[1, 1].set_xlabel("Transmission")
axes[1, 1].set_ylabel("Average Price ($)")

for container in axes[1, 1].containers:
    axes[1, 1].bar_label(
        container,
        fmt="$%.0f",
        padding=3
    )


# ==================== CHART 5 ====================
# Quarterly Revenue Trend

sns.lineplot(
    data=quarterly_revenue,
    x="Quarter",
    y="Revenue",
    hue="Year",
    marker="o",
    linewidth=2.5,
    palette="Dark2",
    ax=axes[2, 0]
)

axes[2, 0].set_title(
    "Quarterly Revenue: 2022 vs 2023",
    fontweight="bold"
)
axes[2, 0].set_xlabel("Quarter")
axes[2, 0].set_ylabel("Revenue ($)")
axes[2, 0].ticklabel_format(style="plain", axis="y")

for _, row in quarterly_revenue.iterrows():
    axes[2, 0].annotate(
        f"${row['Revenue']/1000000:.1f}M",
        (row["Quarter"], row["Revenue"]),
        textcoords="offset points",
        xytext=(0, 8),
        ha="center",
        fontsize=8
    )


# ==================== CHART 6 ====================
# Monthly Sales

sns.barplot(
    data=monthly_sales,
    x="Month",
    y="Sales Count",
    hue="Month",
    palette="Spectral",
    legend=False,
    ax=axes[2, 1]
)

axes[2, 1].set_title(
    "Monthly Car Sales",
    fontweight="bold"
)
axes[2, 1].set_xlabel("Month")
axes[2, 1].set_ylabel("Sales Count")
axes[2, 1].tick_params(axis="x", rotation=45)

for container in axes[2, 1].containers:
    axes[2, 1].bar_label(
        container,
        fmt="%.0f",
        padding=2,
        fontsize=8
    )


plt.tight_layout()
plt.subplots_adjust(hspace=0.35, wspace=0.25)
plt.show()
```

# Insights
- Gender & Body Style: Body style preferences differ between male and female customers.
- Engine: Some engine types are sold more frequently, indicating stronger customer preference.
