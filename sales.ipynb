```python
high5_sales = sales.nlargest(5,"Sales")
high5_sales[["Country","Sales"]]
```

```python
least5_sales = sales.nsmallest(5,"Sales")
least5_sales[["Country","Sales"]]
```

```python
high5_prof = sales.nlargest(5,"Profit")
high5_prof[["Country","Profit"]]
```

```python
least5_prof = sales.nsmallest(5,"Profit")
least5_prof[["Country","Profit"]]
```

```python
product_total = sales.groupby(["Country","Region","Product Name"])["Sales"].sum()
product_total
```
